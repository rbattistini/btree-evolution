```kotlin
%use lets-plot
%use dataframe
```

```kotlin
import java.io.File

fun loadJsonFilesToDataFrame(directoryPath: String): List<File> {
  val dir = File(directoryPath)
  require(dir.isDirectory) { "The provided path is not a directory: $directoryPath" }

  val jsonFiles = dir.listFiles { file -> file.extension == "json" }
  require(!jsonFiles.isNullOrEmpty()) { "No JSON files found in the directory: $directoryPath" }

  return jsonFiles.toList()
}
```

```kotlin
val files = loadJsonFilesToDataFrame("../experiment/archive").map { it.path }
val allExperiments = files.map { DataFrame.read(it) }.reduce { it1, it2 -> it1.concat(it2) }
```

```kotlin
allExperiments
```

```kotlin
val expNames = mapOf(
  "confident_euler" to "black",
  "elegant_lalande" to "magenta",
  "flamboyant_ride" to "blue",
  "lucid_liskov" to "red",
  "inspiring_sanderson" to "green",
  "wonderful_thompson" to "orange"
)
var indexes: MutableMap<Int?, String> = mutableMapOf()
var l = allExperiments.toMap()["name"]
for ((expName, expColor) in expNames) {
  indexes += l?.indexOf(expName) to expColor
}

var paths = emptyList<geomPath>()
for ((idx, color) in indexes) {
  paths += geomPath(size = 0.5, data=allExperiments.generationStats[idx!!].toMap(), color=color, manualKey = allExperiments.toMap()["name"]?.get(idx)) {
    x = "generation"
    y = "bestFitness"
  }
}
var linePlot = ggplot() + ggsize(1200, 600)
for (path in paths) {
  linePlot += path
}
```
